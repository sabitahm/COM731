# Task 1: Load the dataset safely. If the file cannot be found, display a suitable error message.
Use student_scores.csv dataset. 

```python
# Write your code here
import pandas as pd

try:
    student_df = pd.read_csv("student_scores.csv", header=0, encoding = 'utf-8')

except FileNotFoundError:
    print("File not found.")
```

# Task 2: Check whether the column:Mark exists before displaying it

```python
# Write your code here
if "Mark" in student_df.columns:
    print(student_df["Mark"])
else:
    print(list("Column not found."))
```

# Task 3: Display the first two rows only if the DataFrame contains records

```python
# Write your code here
if len(student_df) > 0:
    print(student_df.head(2))
else:
    print("No records found.")
```

# Task 4: Display the last row of the dataset safely and validate the row exists first. Hint: use shape and loc

```python
# Write your code here
num_rows = student_df.shape[0]

if num_rows > 0:
    print(student_df.loc[num_rows -1, :])
else:
    print("No rows found")
```

# Task 5: Create a new CSV file containing only Name and Course. Hint: Check the required columns exist first

```python
# Write your code here
if "Name" in student_df.columns and "Course" in student_df.columns:
    result = student_df[["Name", "Course"]]
    result.to_csv("selected_course.csv", index=False)
else:
    print("Required columns missing.")
```
