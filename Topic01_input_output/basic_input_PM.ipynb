Data Type

```python
#Read in user's data and convert to proper data type and display the result (without formatting output)
weight=float(input("Enter your weight (kg): "))
height=float(input("Enter your height (m): "))
bmi=weight/(height*height)
print(f"Your BMI is: {bmi}")
```

```python
#Read in user's data and convert to proper data type and display the result (with formatting output)
weight=float(input("Enter your weight (kg): "))
height=float(input("Enter your height (m): "))
bmi=weight/(height*height)
print(f"Your BMI is: {bmi:.2f}")
```
