```python
# Define the function (parameter - whatever is inside the brackets)
def welcome(message, moto): # first function parameter and second function parameter
    new_message = message+ ", " +moto
    return (new_message)
# return (new_message)	Sends the result back to wherever the function was called.	
# Without return, the function would give back None and the joined text would be lost.
# Nothing runs at this point. Defining a function only stores the recipe. 
# It runs when you call it.
# Why use return instead of printing inside the function: 
# returning the value lets the caller decide what to do with it, such as printing it, saving it, or passing it to another function. That makes the function more reusable.

# call by its name
msg = input("Enter welcome message: ")
moto = input("Enter a motto: ")
new_welcome_message = welcome(msg, moto) #Calling point of the function
print(new_welcome_message)
```

```python
# Simple sequential version
assessment_details = {} # Dictionary to store assessment details
total_mark = 0

# Get details of assessment
num_assessment = int(input("Enter a number of assessment: "))

for i in range(num_assessment):
    assessment_weight_mark = []
    assessment_type = input("Enter an assessment title [e.g. Ae1, AE2, etc.]")
    weight = int(input("Enter weight of the assessment (e.g. 40, 60 etc.)"))
    weight_percentage = weight/100
    assessment_weight_mark.append(weight_percentage)
    print(f"Assessment_weight_mark: {assessment_weight_mark}")
    assessment_mark = int(input(f"Enter your assessment mark for {assessment_type}"))
    assessment_weight_mark.append(assessment_mark)
    print(f"Assessment_weight_mark: {assessment_weight_mark}")
    assessment_details[assessment_type] = assessment_weight_mark
    print(f"assessment_details: {assessment_details}")

# Calculate the total mark
for key, value in assessment_details.items():
    print(f"Assessment {key}:")
    weight = value[0]
    score = value[1]
    mark = score*weight
    print(f"Mark for {key}: {mark}")
    total_mark = total_mark+mark

print(f"total mark: {total_mark}")


```
