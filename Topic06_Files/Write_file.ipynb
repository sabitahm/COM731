# Write data to a csv file

# write function

```python
save_file_name = 'movies_record.csv'

try:
    with open(save_file_name, 'w', encoding = 'UTF-8') as file:
        print(f"{save_file_name} is open and ready for writing")
        num_movies = int(input("How many movie would you like to record? "))

        file.write("Title, Genre, Rating\n")
        
        for i in range(num_movies):
            title = input("Please enter the title of movie: ")
            genre = input("Please enter the genre of movie: ")
            rating = input("Please enter the rating of movie (1-5): ")

            #write usrer input
            file.write(f"{title}, {genre}, {rating}\n")
            print (f"has been written successfully.")
except Exception as e:
    print(f"An error occured:{e}.")
```

# writer, writerow, writerows from csv module

```python
import csv

account_file = "account.csv"

try:
    with open(account_file, 'w', newline='') as csv_file:
        # create a writer object
        csv_writer = csv.writer(csv_file)

        headers = ['Account', 'Name', 'Balance']
        data = [[1000, 'Jones', 24.98], [200, 'Doe', 345.67], [330, 'White', 25.55]]
        data = []
        row = []
        for i in range(3):
               account_no = int(input("Enter account number: "))
               name = input("Enter a name: ")
               balance = float(input("Enter a balance: "))
               row.append(account_no)
               row.append(name)
               row.append(balance)
               data.append(row)
                
        row = []

        csv_writer.writerow(headers)
        csv_writer.writerows(data)
        print("Your file has been written")
except Exception as e:
    print(f"An error occured: {e}")
```
