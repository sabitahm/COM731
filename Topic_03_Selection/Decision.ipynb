Activity 2: IF-ELIF-ELSE Statement

if_elif_else_statement

```python
#Ask user for the direction
print("Toward which direction should I paint (up, down, left or right)?\n...")

direction = input()

#Direction which message to display
if direction == "up":
    print("I am painting in the upward direction!")
elif direction == "down":
    print("I am paiting in the downward direction!")
elif direction == "left":
    print("I am painting in the leftward direction!")
elif direction == "right":
    print("I am painting in the rightward direction!")
else:
    print("I am not sure which direction to paint!")
```

Activity 3: Modulo Operator

```python
#Ask user for a number 
print("Please enter a whole number\n")

number = int(input())

#Display relevant message
if number % 2 == 0:
    print(f"The number {number} is an even number")
else:
    print(f"The number {number} is an odd number")

```

```python
#Ask user for direction - copied from the solution - extra data to reflect
print("Please enter the first number")
first_number = int(input())

print("Please enter the second number")
second_number = int(input())

#Determine which message to display
if first_number < second_number:
    print("The first number is the smallest")
elif first_number > second_number:
    print("The second number is the smallest")
else:
    print("The two numbers are equal")
```
