#if-elif-else


```python



level = input("Enter your experience level: ")
#check the average saalry
if level.lower() == 'Entry-Level'.lower():
    avg_salary = 39521
elif level == 'Mid-Level':
    avg_salary = 59454
elif level == 'Experienced':
    avg_salary = 80878
else:
    print("Unrecognised experience level")
    avg_salary = 0

if avg_salary == 0:
    print("Average salary can't be identified")
else:
    print(f"The {level} of data scientist earns {avg_salary} per annum")    
```

#Combined condition

```python
iq_range = int(input("Enter an IQ: "))

if iq_range > 140:
    print("Hoghly advanced.")
elif iq_range>=130 and iq_range <=140:
    print("Very advanced.")
elif iq_range>=120 and iq_range <=129:
    print("Superior")
elif iq_range>=110 and iq_range <=119:
    print("High average")
elif iq_range>=90 and iq_range <=109:
    print("Average")
elif iq_range>=80 and iq_range <=89:
    print("Low average")
elif iq_range>=70 and iq_range <=79:
    print("Borderline")
elif iq_range>=69:
    print("Extremely low")
else:
    print("Unidentified IQ classification")
```
