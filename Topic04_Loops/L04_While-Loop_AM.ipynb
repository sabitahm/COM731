While Loop

```python
# Calculate the summation of a series of positive numbers
# receive by a user until the user enter negative number
# (Always to start with 0)
total = 0

number =  int(input("Enter a number or enter negative number to quit: "))

while number >= 0:
    total = total + number
    number = int(input("Enter a number or enter negative number to quit"))

print(f"The summation is {total}")

```

```python
# Calclate the summation of five number recieved by the user (example of infinity loop)

total = 0
count = 0 
while count < 5:
    number = int(input("enter a number or just enter to quit"))
    total = total+number

print(f"The summation is total {total}")

# beocme sinfinity as loop not restricyted
```

```python
# Calclate the summation of five number recieved by the user

total = 0
count = 0 
while count < 5:
    number = int(input("enter a number or just enter to quit"))
    total = total+number
    count = count+1

print(f"The summation is total {total}")

# to make it closed to 5 loop (count = count + 1)
```

```python
#break
# password authentication

correct_password = "secure123"

while True:
    password = input("Enter your password (or exit to quit)")

    if password == 'exit':
        print("Exiting the program...")
        break

    if password == correct_password:
        print("Access granted")
        break

    print("Incorrect password, try again!") 

```

```python
#ask user to enter number between 1 to 5
    
valid_input = False

while valid_input == False:
    try:
        number = int(input("Enter a number (1-5): "))

        if number >= 1 and number <= 5:
            print("Valid number entered.")
            valid_input = True
        else:
            print("Number must be between 1 and 5.")

    except ValueError:
        print("Invalid input. Please enter a whole number.")
```
