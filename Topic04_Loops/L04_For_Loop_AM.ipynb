```python
# Program to find an even number from a range of number
# modulate by 2 is an even number

for i in range(1,31):
    if i%2 == 0:
        print(f"the even number in the range of 1-30: {i}")
```

```python
start = int(input("Enter starting number of range: "))
end = int(input("Enter end umber of range: "))

for number in range (start,end):
    if number%2 == 0:
        print(f"{number} is even number")
        
```

```python
languages = ['Spanish', 'English', 'French', 'German', 'Irish']

for item in languages:
    print(item)
```

```python
#similar to the While loop - using For loop - for input validation (also useful - to repeat the work), 
# but While loop is good to decide when to break, or using more function)
# for loop is conter specific times, diffferent from While loop - control condition to avoid infinite loop iin While)

for count in range(3):
    try:
        mark = int(input(f"Enter mark of test{count+1}: "))
        if mark >=0 and mark <=100:
            print("Valid mark")
        else:
            print("mark must be between 0 and 100")
    except ValueError:
        print("Invalid input. Please enter a whole number")


```
