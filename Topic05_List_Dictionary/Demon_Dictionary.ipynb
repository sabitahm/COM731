```python
scores = {'mandy': 2,
           'fernando': 4, 
           'jessica': 1, 
           'prateek': 1}

print(f"Initialised dictionary: {scores}")
name = input("Enter a name: ")
score = int(input("Enter a score: "))
score[name] = scoreprint(f"The dictionary after adding{name}.{scores})")
```

```python
search_name = input("Enter a name to find: ")
print(f"Score of {search_name}: {scores[search_name]}")
```

```python
try:
    search_name = input("Enter a name to find: ")
    print(f"Score of {search_name}: {scores[search_name]}")
except KeyError:
    print("key is not in a dictionary")
```

```python
# keys () method - return as a list of keys
print(scores.keys())

for keys in scores.keys():
    print(f"Name:{key}, score:{scores[key]}")
```

```python
# Values () method
print(scores.values())

for value in scores.values():
    print(f"Score: {value}")
```

```python
# get method
search_name = input("enter a name: ")

print(f"The score of {search_name} is {scores.get(search_name)}")
```

```python
# delete entry
# first one is delve

print(scores)
delete_name = input("Enter a name to remove: ")

del scores[delete_name]
print(scores)
```

```python
print(scores)
pop_name = input("Enter a name to remove:")
scores.pop(pop_name)
print(scores)
```

```python
#traversing a dictionary 
for item in scores:
    print(item)
```

```python
#traversing a dictionary 
for item in scores:
    print(item, scores[item])
```

```python
print(scores.items())

for key, value in scores.items():
    print(key, value)

    # we cant change the key, but we can change the value of the key)
```
