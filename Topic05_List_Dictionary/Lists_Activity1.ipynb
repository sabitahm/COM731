Activity 2: Indexing 

```python
# Create a list of directions and the number of steps for each movement
paths = ["Move Forward", 10, "Move Backward", 5, "Move Left", 3, "Move Right", 1]

# Let Beep and Bop know that they are starting to move
print("Moving...")

# Display each direction with the number of steps needed
for i in range(0, len(paths), 2):
    direction = paths[i]
    steps = paths[i + 1]
    print(f"{direction} for {steps} steps")
```
